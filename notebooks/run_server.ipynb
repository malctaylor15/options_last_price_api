```python
# This code snippet should be run in a separate cell in your Jupyter Notebook.
# It requires the 'nest_asyncio' and 'uvicorn' libraries.

import uvicorn
import sys
from pathlib import Path

sys.path.append(str(Path.cwd() / "src"))

import nest_asyncio
import threading
import time
```

```python
# Apply nest_asyncio to allow Uvicorn to run within the existing Jupyter event loop
nest_asyncio.apply()

# --- Configuration ---
# Replace 'options_last_price_api.main' with the name of your Python file containing the FastAPI app.
# Replace 'app' with the variable name of your FastAPI instance (e.g., app = FastAPI())
API_MODULE = "options_last_price_api.main:app" 
HOST = "0.0.0.0"
# HOST = "127.0.0.1"
PORT = 9000 # Uvicorn's default port

# Function to run Uvicorn in a separate thread so the notebook cell doesn't block
def start_uvicorn():
    """Starts the Uvicorn server in non-blocking mode."""
    # Note: We omit 'reload=True' here as it can sometimes cause issues in notebooks
    uvicorn.run(
        API_MODULE, 
        host=HOST, 
        port=PORT, 
        log_level="info", 
        factory=False
    )

print(f"Starting FastAPI server on http://{HOST}:{PORT}")
print("Navigate to the URL and add /docs to view the interactive documentation.")

# Run the server in a daemon thread so it stops when the kernel shuts down
server_thread = threading.Thread(target=start_uvicorn, daemon=True)
server_thread.start()

# Wait briefly for the server to spin up
time.sleep(1)

print("\nServer initialization complete. Continue executing other cells or go to /docs.")

# Now you can navigate to http://127.0.0.1:8000/docs in your browser.
```
